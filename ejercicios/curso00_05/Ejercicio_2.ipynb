```python
# Paso 1: Importar las librerías necesarias
import pandas as pd
import matplotlib.pyplot as plt

# Paso 2: Cargar el archivo CSV
# Reemplaza 'tu_archivo.csv' por la ruta o nombre de tu archivo
df = pd.read_csv('hoja.csv')

# Paso 3: Mostrar el DataFrame
# head() muestra las primeras 5 filas del archivo
print("--- Primeras filas del DataFrame ---")
display(df.head())
```

```python
# Paso 4: Mostrar información básica de los datos
print("--- Información general del DataFrame ---")
df.info()

# Paso 5: Crear un gráfico
# Nota: Cambia 'Columna_X' y 'Columna_Y' por los nombres reales de tus columnas
plt.figure(figsize=(10, 6))
plt.plot(df['NUM'], df['CUAD'], marker='o', color='b', linestyle='-')

# Configurar etiquetas y título del gráfico
plt.title('Gráfico de Línea a partir de CSV', fontsize=14)
plt.xlabel('Eje X (NUM)', fontsize=12)
plt.ylabel('Eje Y (CUAD)', fontsize=12)
plt.grid(True)

# Mostrar el gráfico en el cuaderno
plt.show()
```
